```python
"""
Crypto Price Tracker 🪙

This Python script continuously fetches real-time cryptocurrency data from the CoinMarketCap API, ensuring compliance with 
the daily request limit (333 requests per day). It processes the data into a structured format and appends it to a CSV file 
for historical tracking.

Features:
✅ Fetches top 15 cryptocurrencies with their latest prices.
✅ Saves data to a CSV file without overwriting previous entries.
✅ Automatically manages API request limits (333 requests/day).
✅ Uses a .env file to protect API keys.
✅ Implements error handling for reliable execution.
✅ Waits approximately 4.3 minutes between requests to stay within the daily limit.

"""

```

```python
import os
import requests
import pandas as pd
import time
from datetime import datetime
from dotenv import load_dotenv

# Load API Key from .env file
load_dotenv()
API_KEY = os.getenv("CMC_API_KEY")

# API Endpoint
URL = "https://pro-api.coinmarketcap.com/v1/cryptocurrency/listings/latest"
HEADERS = {"X-CMC_PRO_API_KEY": API_KEY}

# Max API calls per day
MAX_REQUESTS_PER_DAY = 333  
INTERVAL = 260  # ~4.3 minutes (ensures we stay within daily limit)

def fetch_crypto_data():
    """Fetch cryptocurrency data from CoinMarketCap API."""
    try:
        response = requests.get(URL, headers=HEADERS, params={"limit": 15, "convert": "USD"})
        response.raise_for_status()
        return response.json().get("data", [])
    except requests.exceptions.RequestException as e:
        print(f"❌ Error fetching data: {e}")
        return []

def save_to_csv(data):
    """Save cryptocurrency data to a CSV file (appends new data)."""
    df = pd.DataFrame([
        {
            "Timestamp": datetime.utcnow().strftime("%Y-%m-%d %H:%M:%S"),
            "Name": crypto["name"],
            "Symbol": crypto["symbol"],
            "Price (USD)": crypto["quote"]["USD"]["price"],
            "Last Updated": crypto["last_updated"]
        }
        for crypto in data
    ])
    
    # Append data to CSV
    file_exists = os.path.isfile("crypto_prices.csv")
    df.to_csv("crypto_prices.csv", mode="a", header=not file_exists, index=False)
    
    print("✅ Data appended to CSV!")

if __name__ == "__main__":
    print("🚀 Crypto tracker started! Press Ctrl+C to stop.")
    request_count = 0
    
    while request_count < MAX_REQUESTS_PER_DAY:
        crypto_data = fetch_crypto_data()
        
        if crypto_data:
            save_to_csv(crypto_data)
            request_count += 1
            print(f"📊 Request {request_count}/{MAX_REQUESTS_PER_DAY} completed.")
        else:
            print("⚠️ No data fetched, skipping this cycle.")
        
        # Wait before next request
        print(f"⏳ Waiting {INTERVAL//60} minutes before next request...\n")
        time.sleep(INTERVAL)

    print("🎯 Daily request limit reached! Script stopped.")
```
